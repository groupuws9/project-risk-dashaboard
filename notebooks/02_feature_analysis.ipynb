# 02 — Feature Analysis (leakage-safe, RQ1)

Mirrors `ml_pipeline/03_feature_engineering.py`.

**Key rule:** raw risk indicators (`employee_turnover_rate`, `absenteeism_rate`, `grievance_rate`, `disputes_raised`, `dispute_risk_level`, `other_industrial_action`, `dismissal_rate`, `disciplinary_warnings`, `suspensions`, `operational_risk_score`) CONSTRUCT the target — they must NOT be model features. Features = organisational characteristics + HR/ER practices + labour-market factors only.

```python
import json, pandas as pd
df = pd.read_csv('../data/processed/wers2011_operational_risk_merged.csv')
meta = json.load(open('../data/processed/dataset_metadata.json'))
cats = meta['variable_categories']
feats = cats['organizational'] + cats['hr_practices'] + cats['employee_relations'] + cats['labor_market']
print(f'{len(feats)} candidate features')
print(df[feats].nunique().sort_values(ascending=False).head(15))
```
