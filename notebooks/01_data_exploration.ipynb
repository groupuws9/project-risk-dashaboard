## What this covers
- Target distribution (Low 5502 / Medium 1105 / High 28) and risk-score descriptives
- Missing values, duplicates, dtypes
- Correlations between organisational / HR features and risk score
- Outputs: `results/eda_report.txt`, `results/summary_statistics.csv`, `results/correlation_matrix.csv`, figures in `results/figures/`

```python
import pandas as pd
df = pd.read_csv('../data/processed/wers2011_operational_risk_merged.csv')
print(df.shape)
print(df['operational_risk_category'].value_counts())
print(df['operational_risk_score'].describe())
```

```python
# Missing values + dtypes
print('missing cells:', int(df.isnull().sum().sum()))
print(df.dtypes.value_counts())
```

```python
# Correlation of numeric features with risk score
num = df.select_dtypes(include='number')
corr = num.corr(numeric_only=True)['operational_risk_score'].drop('operational_risk_score').sort_values(ascending=False)
print(corr.head(15).to_string())
print(corr.tail(10).to_string())
```
